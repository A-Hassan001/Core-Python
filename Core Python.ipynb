# 3. OPERATORS IN PYTHON

An operator is a symbol that performs an operation on one or more operands (values).

Example: a + b, x * y, a > b

Types of Operators
- Arithmetic Operators
- Comparison (Relational) Operators
- Logical Operators
- Assignment Operators
- Identity Operators
- Membership Operators
- Bitwise Operators

**Arithmetic Operators**

| Operator | Name | Example | Meaning |
| --- | --- | --- | --- |
| `+` | Addition | `x + y` | Add two values |
| `-` | Subtraction | `x - y` | Subtract y from x |
| `*` | Multiplication | `x * y` | Multiply two values |
| `/` | Division | `x / y` | Divide x by y (float result) |
| `//` | Floor Division | `x // y` | Divide and get integer result |
| `%` | Modulus | `x % y` | Remainder of division |
| `**` | Exponent | `x ** y` | x raised to power y |

```python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.333...
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

**Comparison (Relational) Operators**

| Operator | Name                     | Example  | Meaning                                 |
| -------- | ------------------------ | -------- | --------------------------------------- |
| `==`     | Equal to                 | `a == b` | True if a is equal to b                 |
| `!=`     | Not equal to             | `a != b` | True if a is not equal to b             |
| `>`      | Greater than             | `a > b`  | True if a is greater than b             |
| `<`      | Less than                | `a < b`  | True if a is less than b                |
| `>=`     | Greater than or equal to | `a >= b` | True if a is greater than or equal to b |
| `<=`     | Less than or equal to    | `a <= b` | True if a is less than or equal to b    |


```python
a = 10
b = 5

print(a == b)  # False
print(a != b)  # True
print(a > b)   # True
print(a < b)   # False
print(a >= b)  # True
print(a <= b)  # False
```

**Logical Operators**

| Operator | Name        | Example   | Meaning                           |
| -------- | ----------- | --------- | --------------------------------- |
| `and`    | Logical AND | `x and y` | True if both x and y are True     |
| `or`     | Logical OR  | `x or y`  | True if any one of x or y is True |
| `not`    | Logical NOT | `not x`   | True if x is False                |


```python
x = True
y = False

print(x and y)  # False
print(x or y)   # True
print(not x)    # False
```

**Assignment Operators**

| Operator | Example   | Same As      |
| -------- | --------- | ------------ |
| `=`      | `x = 5`   | `x = 5`      |
| `+=`     | `x += 3`  | `x = x + 3`  |
| `-=`     | `x -= 3`  | `x = x - 3`  |
| `*=`     | `x *= 3`  | `x = x * 3`  |
| `/=`     | `x /= 3`  | `x = x / 3`  |
| `%=`     | `x %= 3`  | `x = x % 3`  |
| `//=`    | `x //= 3` | `x = x // 3` |
| `**=`    | `x **= 3` | `x = x ** 3` |


```python
x = 10

x += 5    # x = 15
x *= 2    # x = 30
x -= 10   # x = 20
```

## Exercise

Take two numbers from the user and:

1. Perform addition, subtraction, multiplication, division, floor division, modulus, and exponent.
2. Compare the two numbers using comparison operators.
3. Use logical operators to check whether:
4. Both numbers are greater than 0.
5. At least one number is greater than 0.
6. Use an assignment operator to increase the first number by 5.
7. Display all results.

```python
# Get input from user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# 1. Basic Math
print("\n--- Arithmatic Operations ---")
print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Floor Division:", num1 // num2)
print("Modulus (Remainder):", num1 % num2)
print("Exponent (Power):", num1 ** num2)

# 2. Comparisons
print("\n--- Comparisons ---")
print("Is equal?", num1 == num2)
print("Is not equal?", num1 != num2)
print("Is num1 bigger?", num1 > num2)
print("Is num1 smaller?", num1 < num2)
print("Is num1 bigger or equal?", num1 >= num2)
print("Is num1 smaller or equal?", num1 <= num2)

# 3. Logical Checks
print("\n--- Logical Checks ---")
print("Are both positive?", num1 > 0 and num2 > 0)
print("Is at least one positive?", num1 > 0 or num2 > 0)

# 4. Assignment Operator
if num1 > 5:
    num1 += 5
    print("\nFirst number after adding 5:", num1)
else:
    print("First number is less than 5.")
```
